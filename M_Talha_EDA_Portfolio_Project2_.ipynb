# **6. Outlier Detection:**
-  Check for the outliers by using the IQR method.


First Quartile (Q1): The 25th percentile, where 25% of the data lies below this point.

Third Quartile (Q3): The 75th percentile, where 75% of the data lies below this point.

    IQR Formula:
    IQR=Q3−Q1
    IQR=Q3−Q1
        The IQR is the range of values that covers the middle 50% of the dataset.

The IQR helps in understanding how spread out the data is and whether the distribution is concentrated in the center or more dispersed.
    
    Lower Bound: Q1−1.5×IQRQ1−1.5×IQR
    Upper Bound: Q3+1.5×IQRQ3+1.5×IQR

Data points that fall outside these bounds are considered outliers. This helps to highlight extreme values in the data that may require attention or could distort analyses.

```python
treadmill_df1 = treadmill_df.copy()
```

```python
# Calculate the Q1, Q3, IQR
Q3 = treadmill_df1['Income'].quantile(0.75)

Q1 = treadmill_df1['Income'].quantile(0.25)

IQR = Q3 - Q1

# Checking outliers
Lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR

treadmill_df1 = treadmill_df1[(treadmill_df1['Income'] >= Lower_bound) & (treadmill_df1['Income'] <= Upper_bound)]

# Plotting the results
sns.boxplot(x= "Income", data= treadmill_df1)
plt.show()
```

```python
# Calculate the Q1, Q3, IQR
Q3 = treadmill_df1['Miles'].quantile(0.75)

Q1 = treadmill_df1['Miles'].quantile(0.25)

IQR = Q3 - Q1

# Checking outliers
Lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR

treadmill_df1 = treadmill_df1[(treadmill_df1['Miles'] >= Lower_bound) & (treadmill_df1['Miles'] <= Upper_bound)]

# Plotting the results
sns.boxplot(x= "Miles", data= treadmill_df1)
plt.show()
```

# **Conditional Probabilities:**

- What percent of customers have purchased KP281, KP481, or KP781?

```python
Product_percentage = treadmill_df.groupby('Product')[[ "Usage"]].count() / len(treadmill_df)*100
Product_percentage = Product_percentage.round(2).astype(str) + "%"
Product_percentage
```

**Create frequency tables and calculate the percentage as follows**
- **Product – Gender**
 - Percentage of a Male customer purchasing a treadmill
 - Percentage of a Female customer purchasing KP781 treadmill
 - Probability of a customer being a Female given that Product is
   KP281

```python
# Create a crostab function for Product and Gender
Product_Gender = pd.crosstab(treadmill_df['Product'], treadmill_df['Gender'], margins = True)
Product_Gender
```

**Cross-tabulation of products versus gender**

This table shows the number of males and females using each product in the dataset, along with the totals for each product and gender.

1. KP281: 40 females and 40 males use this product, totaling 80 users.

2. KP481: 29 females and 31 males use this product, totaling 60 users.

3. KP781: 7 females and 33 males use this product, totaling 40 users.

```python
# Percentage of a Male customer purchasing a treadmill
Total_male = Product_Gender.loc[:,'Male']
Total_cust = Product_Gender.loc[:,'All']

Percentage_male = (Total_male / Total_cust) * 100
Percentage_male = Percentage_male.round(2).astype(str) + "%"
Percentage_male
```

```python
# Percentage of a Female customer purchasing KP781 treadmill
Female_KP781 = Product_Gender.loc[:,'Female']['KP781']

Percentage_female = (Female_KP781 / Total_cust) * 100
Percentage_female = Percentage_female.round(2).astype(str) + "%"
Percentage_female
```

This question gives us the concept of **Conditional Probability** in which we will define the probability of events:

 1. P(Female) = Customer is Female
 2. P(KP281) = Product is KP281

**The formula of this scenario becomes:**

P(Female|KP281) = P(Female and KP281) / P(KP281)

```python
# Probability of a customer being a Female given that Product is KP281
Prob_female_KP281 = Product_Gender['Female']['KP281'] / Product_Gender['All']['KP281']
Prob_female_KP281 = Prob_female_KP281.round(2).astype(str) + "%"
Prob_female_KP281
```

**Observations:**

1. Female customer prefer to buy KP281 & KP481
2. 50% of female tend to purchase treadmill model KP281
